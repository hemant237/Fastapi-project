```python
# One Hot Encoding
import pandas as pd

df = pd.DataFrame({
    "Color": ["Red", "Blue", "Green", "Red"]
})

df_encoded = pd.get_dummies(
    df,
    columns=["Color"],
    dtype=int
)

print(df_encoded)
```

```python
from sklearn.preprocessing import OrdinalEncoder
import pandas as pd

df = pd.DataFrame({
    "Size":["Small","Medium","Large","Medium","Small"]
})

encoder = OrdinalEncoder(
    categories = [["Small","Medium","Large"]]
)

df["encoded_size"]=encoder.fit_transform(
    df[["Size"]]
)

print(df)
```

```python
import pandas as pd
df = pd.DataFrame({
  "City":["Mumbai","Mumbai","Delhi","Delhi","Pune","Pune"],
  "Salary":[50000,60000,40000,45000,35000,30000]
})

print(df)
```

```python
city_mean = df.groupby("City")["Salary"].mean()
print(city_mean)
```

```python
df["City_encoded"] = df["City"].map(city_mean)
df
```

```python
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df,
    test_size=0.33,
    random_state=42
)

print("Train:")
print(train_df)

print("\nTest:")
print(test_df)
```

```python
city_mean_train = train_df.groupby("City")["Salary"].mean()

print(city_mean_train)
```

```python
train_df["City_encoded"] = train_df["City"].map(
    city_mean_train
)

print(train_df)
```

```python
test_df["City_encoded"] = test_df["City"].map(
    city_mean_train
)

print(test_df)
```

```python
global_mean = train_df["Salary"].mean()

print(global_mean)
```

```python
test_df["City_encoded"] = test_df["City_encoded"].fillna(
    global_mean
)

print(test_df)
```
