```python
import pandas as pd
```

```python
df = pd.DataFrame({
   "height_cm":[170,180,165],
   "weight_kg":[70,80,60] 
})
```

```python
df["BMI"] = df["weight_kg"] / ( (df["height_cm"] / 100) ** 2)

print(df)
```

```python
df = pd.DataFrame({
    "sales":[1000,2000,3000,4000],
    "customers":[10,20,15,40]
})

df["sales_per_cust"] = df["sales"] / df["customers"]

print(df)
```

```python
df = pd.DataFrame({
    "selling_price":[1500,2000,2500,3000],
    "cost_price":[1200,1500,2200,2500]
})

df["profit"] = df["selling_price"] - df["cost_price"]

print(df)
```

```python
#Interaction Feature

df =pd.DataFrame({
    "experience":[1,2,5,10],
    "education":[1,3,2,5]
})

df["experience_education"] = df["experience"] * df["education"]

print(df)
```

```python
df = pd.DataFrame({
    "date":pd.to_datetime([
        "2026-01-13",
        "2026-03-25",
        "2026-05-9",
        "2026-12-26"
    ])
})

df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day
df["day_of_week"] = df["date"].dt.dayofweek

print(df)
```
