```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "income": [
        20000,
        25000,
        30000,
        40000,
        50000,
        75000,
        100000,
        200000,
        500000
    ]
})

df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.hist(df["income"],bins=20)

plt.xlabel("Income")
plt.ylabel("Frequency")
plt.title("Income Distribution")

plt.show()
```

```python
df["log_income"] = np.log1p(df["income"])
```

```python
print(df)
```

```python
plt.figure(figsize=(6,4))

plt.hist(df["log_income"],bins=8)

plt.xlabel("Log Income")
plt.ylabel("Frequency")
plt.title("Income Log Transformed Distribution")

plt.show()
```
