```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("train.csv")
```

```python
print(df.columns)
```

```python
# Coorelation with Price
coorelation = df[
    ["Distance_to_Metro_km","Distance_to_City_Center_km","Price_INR_Lakhs"]
].corr()

print("Coorelation Matrix")
print(coorelation)
```

```python
# Distance to metro vs price
plt.figure(figsize=(8,4))

plt.scatter(
    df["Distance_to_Metro_km"],
    df["Price_INR_Lakhs"],
    alpha=0.3
)

plt.xlabel("Distance to Metro")
plt.ylabel("Price INR In Lakhs")
plt.title("Distance  to Metro VS Price Distribution")
plt.tight_layout()
plt.show()
```

```python
# Distance to city centre vs Price

plt.figure(figsize=(7,4))

plt.scatter(
    df["Distance_to_City_Center_km"],
    df["Price_INR_Lakhs"],
    alpha=0.3
)

plt.xlabel("Distance to city centre")
plt.ylabel("Price in Lakhs")
plt.title("Distance to city centre vs Price Distribution")
plt.tight_layout()
plt.show()
```
