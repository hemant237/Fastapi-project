```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("train.csv")

# Average and median price by bathrooms
bathroom_price = (
    df.groupby("Bathrooms")["Price_INR_Lakhs"]
    .agg(["mean", "median", "count"])
)

print("Price Analysis by Bathrooms:")
print(bathroom_price)

# Average price visualization
plt.figure(figsize=(8, 5))
bathroom_price["mean"].plot(kind="bar")

plt.xlabel("Bathrooms")
plt.ylabel("Average Price (INR Lakhs)")
plt.title("Average Property Price by Number of Bathrooms")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
