### Predict prices for test.csv and create the submission file

```python
import joblib
import pandas as pd

model = joblib.load("../models/hist_gb_tuned.joblib")

test_df = pd.read_csv("../test.csv")
X_test = test_df.drop(columns=["Property_ID"])

submission = pd.DataFrame({
    "Property_ID": test_df["Property_ID"],
    "Price_INR_Lakhs": model.predict(X_test).round(2)
})

submission.to_csv("../submission.csv", index=False)

print(submission.shape)
submission.head()
```
