```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../train.csv")
```

```python
#Target Column
target = df["Price_INR_Lakhs"]
```

```python
# Basic Statistics
print("\nMean Price")
print(target.mean())

print("\nMedian Price")
print(target.median())

print("\nStandard Deviation")
print(target.std())

print("\nMaximun Price")
print(target.max())

print("\nMinimum Price")
print(target.min())

print("\nSkewness")
print(target.skew())
```

```python
# Histogram
plt.figure(figsize=(8,4))

plt.hist(
    target,
    bins=50
)

plt.xlabel("Price INR Lakhs")
plt.ylabel("Frequency")
plt.title("Price Distribution")

plt.show()
```

```python
plt.figure(figsize=(10,4))

plt.boxplot(target)

plt.xlabel("Price INR Lakhs")
plt.title("Box Plot of House Price")

plt.show()
```
