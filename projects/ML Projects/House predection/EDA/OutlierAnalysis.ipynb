```python
import pandas as pd
```

```python
df = pd.read_csv("../train.csv")
```

```python
price = df["Price_INR_Lakhs"]
```

```python
#Calculate Quartiles
Q1 = price.quantile(0.25)
Q3 = price.quantile(0.75)

IQR = Q3-Q1
```

```python
# Calculate Boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)

print("\nLower Bound")
print(lower_bound)

print("Upper Bound")
print(upper_bound)
```

```python
# Identify Outliers
outliers = df[(price < lower_bound) | (price > upper_bound)]

print("\nNo of Outliers")
print(len(outliers))

print("\nPercentage of Outliers")
print(len(outliers)/ len(df) * 100)
```

```python
# Display Extreme Prices

print("\n Highest 10 prices")
print(df["Price_INR_Lakhs"].sort_values(ascending=False).head(10))
```
