```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../train.csv")
```

```python
# Average and median price by BHK
bhk_price = df.groupby("BHK")["Price_INR_Lakhs"].agg(["mean","median","count"])

print("\nAverage price by BHK")
print(bhk_price)
```

```python
# Average price visualization
plt.figure(figsize=(8,4))

bhk_price["mean"].plot(kind="bar")

plt.xlabel("BHK")
plt.ylabel("Avg Price")
plt.title("Average price vs BHK")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
