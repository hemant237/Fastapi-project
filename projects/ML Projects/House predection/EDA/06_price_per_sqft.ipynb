```python
import pandas as pd


# ==========================================
# 1. Load Dataset
# ==========================================

df = pd.read_csv("../train.csv")
```

```python
# 2. Create Price Per SqFt
df["Price_per_sqft"]= df["Price_INR_Lakhs"] * 100000 / df["Super_Area_SqFt"]
```

```python
# 3. Display Results
print(df[
    [
         "City",
            "Super_Area_SqFt",
            "Price_INR_Lakhs",
            "Price_per_sqft"
    ]
].head(10))
```
