```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../train.csv")

# Average and median price by property age
age_price = (
    df.groupby("Age_of_Property")["Price_INR_Lakhs"]
    .agg(["mean", "median", "count"])
)

print("Price Analysis by Property Age:")
print(age_price)

# Correlation
correlation = df["Age_of_Property"].corr(df["Price_INR_Lakhs"])

print("\nCorrelation between Age and Price:")
print(correlation)

# Scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(
    df["Age_of_Property"],
    df["Price_INR_Lakhs"],
    alpha=0.3
)

plt.xlabel("Age of Property (Years)")
plt.ylabel("Price (INR Lakhs)")
plt.title("Property Age vs Price")

plt.tight_layout()
plt.show()
```
