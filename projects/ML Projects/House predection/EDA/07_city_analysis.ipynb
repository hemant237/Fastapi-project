```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Load Dataset
# ==========================================

df = pd.read_csv("../train.csv")
```

```python
#Average Price by City
city_price =  ( df.groupby("City")["Price_INR_Lakhs"].mean().sort_values(ascending=False) )

print("Average City By Price")
print(city_price)
```

```python
#median Price by City
median_price = ( df.groupby("City")["Price_INR_Lakhs"].median().sort_values(ascending=False) )

print("\nMedian Price by City")
print(median_price)
```
