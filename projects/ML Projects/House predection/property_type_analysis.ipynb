```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("train.csv")
```

```python
# Average price by property type
property_avg = df.groupby('Property_Type')["Price_INR_Lakhs"].mean().sort_values(ascending=False)

print("\nAverage Price by Property Type")
print(property_avg)
```

```python
# Median price
median_price = df.groupby('Property_Type')["Price_INR_Lakhs"].median().sort_values(ascending=False)

print("\nMedian Price by Property Type")
print(median_price)
```

```python
# Number of properties
prop_count = df["Property_Type"].value_counts().sort_values(ascending=False)

print("\nproperty Count by Type")
print(prop_count)
```

```python
# Visualization
plt.figure(figsize=(8,4))

property_avg.plot(kind="bar")

plt.xlabel("Property Type")
plt.ylabel("Avg Price INR Lakhs")
plt.title("Property Type Count")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```
